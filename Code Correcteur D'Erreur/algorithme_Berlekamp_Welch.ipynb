```python
#Parametres du code
q = 7
F = GF(q)
R_poly.<X> = PolynomialRing(F)

n = 5
k = 3
t = (n-k)/2

#Les points d'evaluation des x
x_points = [F(i) for i in range(n)]

#Generation d'un message (polynome de degre < k)
f_orig = R_poly.random_element(degree=k-1)
print(f"Polynome original f(X):{f_orig}")

#Calcul du mot de code de f
motCode = [f_orig(xi) for xi in x_points]

#Simulation d'Erreurs de poids inferieurs a <=t
u = list(motCode)
errors_indices = [0, 2, 5, 10]
for i in errors_indices[:t]:
    u[i] += F.random_element()

print(f"Nombre d'erreurs introduites:{min(len(errors_indices), t)}")
```

```python
rows = []
for i in range(n):
    row_A = [u[i]*(x_points[i]**j) for j in range(t+1)]
    row_B = [-(x_points[i]**j) for j in range(n-t)]
    rows.append(row_A+row_B)
M = matrix(F, rows)

sol_vector = M.right_kernel().basis()[0]
print(f"la solution non trivial est: {sol_vector}")

a_coeffs = sol_vector[:t+1]
b_coeffs = sol_vector[t+1:]

A_poly = R_poly(list(a_coeffs))
B_poly = R_poly(list(b_coeffs))

print(f"A(X)={A_poly}")
print(f"B(X)={B_poly}")
```

```python
#Algorithme de Decodage

f_dec, rest = B_poly.quo_rem(A_poly)

if rest == 0:
    print(f"Decodage du message f: {f_dec}")
    print(f"Verification:{f_dec}=={f_orig}")
else:
    print("Erreur de decodage")
```
