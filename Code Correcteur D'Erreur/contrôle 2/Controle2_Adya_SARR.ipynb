```python
def phi_tild(x_family, beta):
    F_ext = beta.parent()
    m = F_ext.degree()
    
    all_columns = []
    
    for x_j in x_family:
        column_j = []
        for element in x_j:
            coords = element.polynomial().list()
            coords += [0] * (m - len(coords))
            column_j.extend(coords)
        all_columns.append(vector(GF(2), column_j))
    return matrix(GF(2), all_columns).transpose()
v1 = vector(F_16, [1, beta])
v2 = vector(F_16, [beta^2, 0])
v3 = vector(F_16, [beta^2, beta])
x_family = [v1, v2, v3]

M_phi = phi_tild(x_family, beta)
print(M_phi)
```

```python
g_vec = [beta, beta^2, beta^3, beta^4, beta^5]

def get_rank_matrix(vec_list, m=4):
    columns = []
    for el in vec_list:
        # Extraction des coefficients du polynôme (a0, a1, a2, a3)
        coords = el.polynomial().list()
        coords += [0] * (m - len(coords)) # Complétion à 4 coordonnées
        columns.append(vector(GF(2), coords))
    return matrix(GF(2), columns).transpose()

print(get_rank_matrix(g_vec))
```
