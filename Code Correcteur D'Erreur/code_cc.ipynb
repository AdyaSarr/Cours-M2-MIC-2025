```python
F = GF(7)
R.<X> = PolynomialRing(F)

x_pts = [F(0), F(1), F(2), F(3), F(4)]  
u = vector(F, [1, 6, 3, 0, 3])
n = 5
k = 3
t = (n - k) // 2
var_names = [f'a{i}' for i in range(t + k)] + ['b0']
V = FreeModule(F, len(var_names))
matrix_rows = []
constants = []

for i in range(n):
    xi = x_pts[i]
    ui = u[i]
    row = [xi^0, xi^1, xi^2, xi^3, -ui]
    matrix_rows.append(row)
    constants.append(ui * xi)

M = matrix(F, matrix_rows)
Y = vector(F, constants)

sol = M.solve_right(Y)


A = R([sol[0], sol[1], sol[2], sol[3]])
B = X + sol[4]


f_recupere = A // B

print(f"A(X) = {A}")
print(f"B(X) = {B}")
print(f"Polynôme de message f(X) retrouvé : {f_recupere}")
```
