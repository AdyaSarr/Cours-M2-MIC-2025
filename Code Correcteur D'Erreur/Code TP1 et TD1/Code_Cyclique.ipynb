```python
# Paramètres : Corps F2, longueur n=7 (comme pour Hamming 7,4)
n = 7
F = GF(2)
R.<X> = PolynomialRing(F)

# Pour un code cyclique, g(X) doit diviser X^n - 1
# Choisissons g(X) = X^3 + X + 1 (générateur du Hamming cyclique)
g = X^3 + X + 1
target = X^n - 1

if target % g == 0:
    print(f"g(X) est un polynôme générateur valide car il divise X^{n}-1")
else:
    raise ValueError("g(X) ne divise pas X^n - 1")

# Calcul de la dimension k = n - deg(g)
k = n - g.degree() # 7 - 3 = 4
```

```python
def codage_cyclique_systematique(message_bits):
    """
    Code un message m(X) de longueur k de manière systématique.
    Résultat : c(X) = m(X)*X^(n-k) - (m(X)*X^(n-k) mod g(X))
    """
    m_poly = R(list(message_bits))
    r = n - k
    
    # Décalage du message : m(X) * X^r
    shifted_m = m_poly * X^r
    
    # Calcul du reste (redondance)
    rem = shifted_m % g
    
    # Mot de code final
    c_poly = shifted_m - rem
    
    # On s'assure que le vecteur a une longueur n
    coeffs = c_poly.list()
    return coeffs + [0]*(n - len(coeffs))

# Test codage
msg = [1, 0, 1, 1]
mot_code = codage_cyclique_systematique(msg)
print(f"Message : {msg} -> Mot de code : {mot_code}")
```

```python
def decodage_cyclique(y_recu):
    """
    Calcule le syndrome polynomial s(X) = y(X) mod g(X).
    """
    y_poly = R(list(y_recu))
    syndrome = y_poly % g
    
    if syndrome == 0:
        print("Aucune erreur détectée.")
        # Extraction du message (pour le systématique, ce sont les k derniers bits)
        return list(y_recu)[n-k:]
    else:
        print(f"Erreur détectée ! Syndrome : {syndrome}")
        # Note : Pour corriger, il faudrait une table de syndromes ou 
        # l'algorithme de Meggitt (plus complexe).
        return None

# Simulation
y_errone = list(mot_code)
y_errone[2] = (y_errone[2] + 1) % 2 # Ajout d'une erreur
decodage_cyclique(y_errone)
```
